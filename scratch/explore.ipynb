```python
import pandas as pd
```

```python
features = pd.read_csv("features.csv")
```

```python
features.columns
```

```python
features["CellType"] = features["Image Path"].apply(lambda x: x.split("/")[-2])
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(
    data=features,
    x="Cell_MajorAxisLength",
    hue="CellType",
    bins=200,            # ↑ more steps (increase as needed)
    element="step",      # outline instead of bars
    fill=False,          # remove shading
    stat="density",      # optional
    common_norm=False
)
plt.xlim(0, 200)
plt.show()
```
