```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
features = pd.read_csv("features_2.csv")
```

```python
features.columns
```

```python
features["SegmentationMethod"].value_counts()
```

```python
features["CellType"] = features["Image Path"].apply(lambda x: x.split("/")[-2])
```

```python
sns.histplot(
    data=features[features["SegmentationMethod"] == "watershed"],
    x="CellMeanIntensity_PE",
    hue="CellType",
    bins=1000,            # ↑ more steps (increase as needed)
    element="step",      # outline instead of bars
    fill=False,          # remove shading
    stat="density",      # optional
    common_norm=False
)
plt.xlim(0, 3000)
plt.show()
```
